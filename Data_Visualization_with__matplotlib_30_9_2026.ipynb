```python
import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
df = pd.read_csv('/content/web_traffic_daily.csv',parse_dates=["date"])
df.info()
print("not missing  values")
df.notnull()
df.head()
print("Duplicate values")
df.duplicated()
df.shape
print("the column names:",df.columns)

```

how did the no.of visitors of our website change over year??

```python
import matplotlib.pyplot as plt

# Create figure and axes with the desired size
fig, ax = plt.subplots(figsize=(7, 4.2))

# Plot data
ax.plot(df['date'], df['visits'], color='tab:blue', linewidth=1.5, label='Visits')

# Set titles and labels
plt.title("Daily visits over one year")
plt.xlabel("Date")
plt.ylabel("Visits")

# Customize grid and display
plt.grid(alpha=0.3)
plt.show()
```

```python
fig ,ax = plt.subplots(figsize = (10,4))

ax.plot(
    df["date"],
    df["visits"],
    color = "tab:blue",
    linewidth = 1.5,
    linestyle = "-",
    label = "Visits"
)

ax.plot(
    df["date"],
    df["signups"],
    color="tab:orange",
    linewidth = 1.5,
    linestyle = "--",
    label = "Signups"
)
ax.set_title("Daily Visits and Signups Over one year")
ax.set_xlabel("Done")
ax.set_ylabel("Count")
ax.set_yscale("log")
ax.legend()
ax.grid(alpha = 0.3)

plt.tight_layout()
plt.show()
```

Does a slower website experience more errors, how does website traffic is related to these errors?


```python
import matplotlib.pyplot as plt
plt.figure(figsize = (10,4))
sc = plt.scatter(df["page_load_ms"] , df["error_count"])
plt.title("Error vs pageload time")
plt.xlabel("page load time(ms)")
plt.ylabel("Error count")
plt.colorbar(sc, label = "visits")
plt.show()
```

```python
monthly = df.set_index("date").resample("ME")[["visits", "signups"]].sum()

monthly.index = monthly.index.strftime("%b")

plt.figure(figsize=(10, 4))

x = np.arange(len(monthly))
width = 0.35

plt.bar(
    x - width/2,
    monthly["visits"],
    width,
    label="Visits"
)

plt.bar(
    x + width/2,
    monthly["signups"] * 10,
    width,
    label="Signups (x10 for scale)"
)

plt.xticks(x, monthly.index, rotation=45)

plt.title("Monthly Visits vs Signups (Grouped Bar)")
plt.xlabel("Month")
plt.ylabel("Count")
plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.show()
```
