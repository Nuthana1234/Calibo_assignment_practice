```python
import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
df = pd.read_csv('/content/web_traffic_daily.csv',parse_dates=["date"])
df.info()
print("not missing  values")
df.notnull()
df.head()
print("Duplicate values")
df.duplicated()
df.shape
print("the column names:",df.columns)

```

how did the no.of visitors of our website change over year??

```python
import matplotlib.pyplot as plt

# Create figure and axes with the desired size
fig, ax = plt.subplots(figsize=(7, 4.2))

# Plot data
ax.plot(df['date'], df['visits'], color='tab:blue', linewidth=1.5, label='Visits')

# Set titles and labels
plt.title("Daily visits over one year")
plt.xlabel("Date")
plt.ylabel("Visits")

# Customize grid and display
plt.grid(alpha=0.3)
plt.show()
```

```python
fig ,ax = plt.subplots(figsize = (10,4))

ax.plot(
    df["date"],
    df["visits"],
    color = "tab:blue",
    linewidth = 1.5,
    linestyle = "-",
    label = "Visits"
)

ax.plot(
    df["date"],
    df["signups"],
    color="tab:orange",
    linewidth = 1.5,
    linestyle = "--",
    label = "Signups"
)
ax.set_title("Daily Visits and Signups Over one year")
ax.set_xlabel("Done")
ax.set_ylabel("Count")
ax.set_yscale("log")
ax.legend()
ax.grid(alpha = 0.3)

plt.tight_layout()
plt.show()
```

Does a slower website experience more errors, how does website traffic is related to these errors?


```python
import matplotlib.pyplot as plt
plt.figure(figsize = (10,4))
sc = plt.scatter(df["page_load_ms"] , df["error_count"])
plt.title("Error vs pageload time")
plt.xlabel("page load time(ms)")
plt.ylabel("Error count")
plt.colorbar(sc, label = "visits")
plt.show()
```

```python
monthly = df.set_index("date").resample("ME")[["visits", "signups"]].sum()

monthly.index = monthly.index.strftime("%b")

plt.figure(figsize=(10, 4))

x = np.arange(len(monthly))
width = 0.35

plt.bar(
    x - width/2,
    monthly["visits"],
    width,
    label="Visits"
)

plt.bar(
    x + width/2,
    monthly["signups"] * 10,
    width,
    label="Signups (x10 for scale)"
)

plt.xticks(x, monthly.index, rotation=45)

plt.title("Monthly Visits vs Signups (Grouped Bar)")
plt.xlabel("Month")
plt.ylabel("Count")
plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
import matplotlib.pyplot as plt

# Set the figure size
plt.figure(figsize=(7, 4))

# Pass the dataframe column first, then the remaining plot styling arguments
plt.hist(df["page_load_ms"], bins=20, color="tab:orange", edgecolor="black")

# Add titles and labels
plt.title("DISTRIBUTION OF PAGE LOAD TIMES")
plt.xlabel("page load time(ms)")
plt.ylabel("no.of days")

# Render the plot
plt.show()
```

```python
deploy_days = df["deployment_flag"].sum()
normal_days = len(df) - deploy_days

plt.figure(figsize = (5,5))
plt.pie([deploy_days , normal_days],
        labels = ["Deployment day","Normal day"],
        autopct = "%1.1f%%" , startangle = 90)
plt.title("Share of deployment days in the year")
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(11, 7), constrained_layout=True)

axes[0, 0].plot(df["date"], df["visits"], color="tab:blue")
axes[0, 0].set_title("Visits over time")

axes[0, 1].hist(df["page_load_ms"], bins=25, color="tab:orange")
axes[0, 1].set_title("Page load distribution")

axes[1, 0].scatter(df["visits"], df["signups"], alpha=0.5, color="tab:green")
axes[1, 0].set_title("Visits vs Signups")

axes[1, 1].boxplot(df["error_count"])
axes[1, 1].set_title("Error count spread")

fig.suptitle("Website health dashboard", fontsize=14)
plt.show()

# Try this: change to plt.subplots(1, 4, figsize=(18, 4)) and index axes as axes[0], axes[1], ...
```

```python
df['visits_7d_avg'] = df['visits'].rolling(7).mean()
incident_date = df.loc[df["error_count"].idxmax() ,"date"]

fig, ax = plt.subplots(figsize = (11 ,4.5))
ax.plot(df["date"],df["visits"],alpha = 0.35,label = "Daily Visits")
ax.plot(df["date"],df["visits"],color = "tab:blue",linewidth = 2, label = "7-day average")
ax.axvline(incident_date , color = "red",linestyle = "--",label = "Worst incident day")
ax.set_title("visits with 7-day rolling average and incident marker")
ax.legend()
plt.show()
```

```python
fig, ax1 = plt.subplots(figsize=(11, 4.5))

ax1.plot(df["date"], df["page_load_ms"], color="tab:red")
ax1.set_ylabel("Page load (ms)", color="tab:red")
ax1.tick_params(axis="y", labelcolor="tab:red")

ax2 = ax1.twinx()
ax2.plot(df["date"], df["error_count"], color="tab:gray", alpha=0.6)
ax2.set_ylabel("Error count", color="tab:gray")
ax2.tick_params(axis="y", labelcolor="tab:gray")

ax1.set_title("Page load time and error count share one x-axis, different y-scales")
plt.show()
```

```python
import seaborn as sns
import pandas as pd
import numpy as np
print("seaborn version check:" ,sns.__version__)
print("Pandas version check:",pd.__version__)
print("numpy version check:",np.__version__)
```

```python
pf = pd.read_csv("/content/student_tech_profile.csv")
pf.head()
```

```python
pf.describe()
```

```python
sns.set_theme(style = "whitegrid", palette = "muted")
sns.set_context("notebook")
fig ,ax = plt.subplots(figsize = (6,4))
sns.histplot(data = pf , x="cgpa",ax =ax)
ax.set_title("Seaborn theme applied ,still a matplotlib axes underneath")
plt.show()
```

```python
fig,ax = plt.subplots(figsize = (7,4))
sns.histplot(data = pf , x = "coding_hours_per_week",kde = True , ax = ax)
ax.set_title("Distribution of weekly coding hours")
plt.show()
```

```python
fig,ax = plt.subplots(figsize = (7,4))
sns.kdeplot(data = pf , x = "cgpa",hue = "placed",fill = True, alpha = 0.4 ,ax=ax)
ax.set_title("CGPA distribution :placed vs non-placed")
plt.show()
```

What if I want to compare the coding hour distribution seperately for eachengineering branch?

```python
sns.displot(data = pf ,x = "coding_hours_per_week",col = "branch",col_wrap = 3, height = 2.8)
plt.suptitle("coding hours distribution per branch(figure - level, fafaceted)",y = 1.02)
plt.show()
```

ECDF = Emperical Cumilative Distribution Function

tells the percentage or proportion of observations

```python
fig ,ax = plt.subplots(figsize = (7,4))
sns.ecdfplot(data = pf, x = "cgpa",hue = "placed",ax = ax)
ax.set_title("What fraction of students have cgpa  < =x?")
plt.show()
```
