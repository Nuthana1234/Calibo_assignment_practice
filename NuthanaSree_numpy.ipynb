1.   NumPy arrays are preferred over Python lists because they store data in contiguous memory blocks and use a single, uniform data type.
2.   Vectorization is the process of performing mathematical operations on entire arrays at once, entirely eliminating explicit for-loops in Python.


1. A * B (Element-wise Multiplication): Multiplies corresponding elements of arrays A and B. The arrays must have the exact same shape.
 A @ B (Matrix Multiplication / Dot Product): Performs standard linear algebra matrix multiplication where row elements of A are multiplied by column elements of B and summed. The inner dimensions must match.
2. Broadcasting is a mechanism that allows NumPy to perform arithmetic operations on arrays of different shapes without making unnecessary, memory-expensive copies of the data in memory.


1. flatten(): Always returns a copy of the original array collapsed into a single dimension. Modifying a flattened array does not alter the original data.ravel(): Returns a flattened, contiguous 1D view of the original array whenever possible without allocating new memory. Modifying a raveled array will modify the original data.



1. axis=0 (Column-wise): The operation collapses the rows. It moves down the columns, calculating a result for each column.
  axis=1 (Row-wise): The operation collapses the columns. It moves across the rows, calculating a result for each row.
  
  Code:

import numpy as np

arr = np.array([[1, 2],
                [3, 4]])

print(np.sum(arr, axis=0))  # Output: [4, 6]

print(np.sum(arr, axis=1))  # Output: [3, 7]

1.Given a 1D NumPy array of exam scores, use np.where() to produce labels 'P' (score >= 40) or 'F' (score < 40). [5 marks]​

2.Write a function standardize(x) that takes a 1D NumPy array and returns its z-score standardized version using only NumPy. [5
marks]​

```python
import numpy as np

def standardize(x):
    return (x - np.mean(x)) / np.std(x)
data = np.array([10, 20, 30, 40, 50])
print(standardize(data))
```

4.Given X with shape (100, 5), compute column-wise mean and standard deviation and standardize the matrix in one vectorized
expression. [5 marks]​

```python
import numpy as np

# Generate dummy matrix X with shape (100, 5)
X = np.random.rand(100, 5)

# Vectorized standardization along columns (axis=0)
X_standardized = (X - np.mean(X, axis=0)) / np.std(X, axis=0)

print("Standardized shape:", X_standardized.shape)
```

5.Implement y_pred = X @ w + b for a feature matrix X, weight vector w and bias b. Then compute MSE using y_true. [5 marks]​

```python
import numpy as np
# Feature matrix
X = np.array([
    [1, 2],
    [2, 3],
    [3, 4],
    [4, 5]
])
# Weight vector
w = np.array([2, 3])
# Bias
b = 1
# True values
y_true = np.array([9, 14, 19, 24])
# Prediction
y_pred = X @ w + b
# Mean Squared Error
mse = np.mean((y_true - y_pred) ** 2)
print("Predicted values:", y_pred)
print("MSE:", mse)
```

7.Given a grayscale image as a 2D NumPy array with pixel values in [0, 255], normalize the values to [0, 1] using min-max
normalization. [5 marks]​

```python
import numpy as np
# Sample grayscale image
image = np.array([
    [0, 50, 100],
    [150, 200, 255]
])
# Min-max normalization
image_normalized = (image - image.min()) / (image.max() - image.min())
print("Original image:")
print(image)
print("\nNormalized image:")
print(image_normalized)
```

1.You are given a raw 1D or 2D NumPy array containing NaN values.​

Describe, step by step, the NumPy operations you would apply to:​

(a) detect missing values,​

(b) handle them appropriately and justify your strategy, and​

(c) normalize or standardize the clean data before it is fed to a model.​

​

Implement the full pipeline in code on a sample array of your own choosing, including at least one NaN.​

```python
import numpy as np
#Sample 2D array containing NaN values
data = np.array([
    [10.0, 20.0, np.nan],
    [30.0, 40.0, 50.0],
    [60.0, np.nan, 80.0]
])
#Detect missing values
missing_values = np.isnan(data)
print("Missing values:")
print(missing_values)
print("Number of NaN values:", np.isnan(data).sum())
#Handle missing values using mean imputation
mean_value = np.nanmean(data)
data_clean = np.nan_to_num(data, nan=mean_value)
print("\nMean used for imputation:", mean_value)
print("Clean data:")
print(data_clean)
# Min-max normalization
data_normalized = (
    (data_clean - data_clean.min()) /
    (data_clean.max() - data_clean.min())
)

print("\nNormalized data:")
print(data_normalized)
```

E1. Per-Column Min–Max Normalization [5 marks]​

Implement min-max normalization for a 2D feature matrix on a per-column (per-feature) basis, rather than using a single global
min/max for the whole array. Your solution must be fully vectorized, with no explicit loops over rows or columns.​

```python
import numpy as np

X = np.array([
    [10, 100, 1000],
    [20, 200, 2000],
    [30, 300, 3000],
    [40, 400, 4000]
])

# Per-column minimum and maximum
X_min = X.min(axis=0)
X_max = X.max(axis=0)

# Per-column Min-Max normalization
X_normalized = (X - X_min) / (X_max - X_min)

print("Original matrix:")
print(X)

print("\nNormalized matrix:")
print(X_normalized)
```

E2. Three-Level Score Classification [5 marks]​

Given a 1D array of exam scores, use a single vectorized expression (no if/else statements and no explicit loops) to classify each
score as:​

Fail (< 40), Average (40–70 inclusive), or Excellent (> 70).​

Hint: You may need to nest or combine np.where() calls.​

```python
import numpy as np

scores = np.array([25, 40, 55, 70, 85, 95])

result = np.where(
    scores < 40,
    "Fail",
    np.where(scores <= 70, "Average", "Excellent")
)
print("Result")
print(result)
```
