Q2. Data Analysis​


Issue: sort_values('Salary') sorts in ascending order by default, so .head(5) gives the five lowest salaries.

```python
import pandas as pd

df = pd.DataFrame({
    'Employee': ['A', 'B', 'C', 'D', 'E', 'F'],
    'Salary': [30000, 50000, 25000, 60000, 45000, 70000]
})

print(df.sort_values('Salary', ascending=False).head(5))
```

Explanation: ascending=False sorts salaries from highest to lowest, so head(5) returns the five highest salaries.

Q3. Cleaning Data

Issue: fillna() returns a modified Series but does not automatically update the original DataFrame.

```python
import pandas as pd

df = pd.DataFrame({'Marks': [70, None, 90]})

df['Marks'] = df['Marks'].fillna(df['Marks'].mean())

print(df)
```

Explanation: The missing value is replaced with the column mean, and the result is assigned back to df['Marks'].

Q1. NumPy Broadcasting

```python
import numpy as np
a = np.array([[10], [20], [30]])
b = np.array([[1, 2, 3]])
result = a + b
print(result)
print(result.shape)
```

Explanation: Broadcasting expands the smaller dimensions so the arrays can be added element-wise.

If b = np.array([1, 2]), a broadcasting error occurs because the dimensions are incompatible.

Q3. Extreme Salary

```python
import pandas as pd

df = pd.DataFrame({
    "Employee": ["A", "B", "C", "D"],
    "Salary": [25000, 30000, 35000, 10000000]
})

print("Mean:", df["Salary"].mean())
print("Median:", df["Salary"].median())
```

The extreme salary of 10,000,000 significantly increases the mean. The median is less affected and better represents the typical salary in this dataset.

The record should be investigated before removing it because the extreme value may be genuine or may indicate a data-entry error

Q1. Cleaning Sequence

```python
import pandas as pd
df = pd.read_csv("/content/student_assessment_dirty.csv")

# Inspect
print(df.head())
df.info()
print(df.isnull().sum())
# Standardize departments
df["Department"] = (
    df["Department"]
    .str.strip()
    .str.upper()
    .str.replace(".", "", regex=False)
)
# Remove exact duplicates
before = len(df)
df = df.drop_duplicates()
records_removed = before - len(df)
# Convert marks to numeric
df["Marks"] = (
    df["Marks"]
    .astype("string")
    .str.extract(r"(\d+(?:\.\d+)?)")[0]
)

df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

# Identify missing marks
missing_marks = df["Marks"].isna().sum()

# Identify invalid marks
invalid_marks = df[
    df["Marks"].notna() &
    ~df["Marks"].between(0, 100)
]

df
print("Records removed:", records_removed)
print("Missing marks:", missing_marks)
print("Invalid marks:")
print(invalid_marks)
```
