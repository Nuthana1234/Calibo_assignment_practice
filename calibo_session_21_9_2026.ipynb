### 🎯 Your Task:
Using **Pandas Series**, perform the following:
✅ Create a Series named `scores` using the dictionary `student_scores`.
✅ Print the marks scored by **"Charlie"** and **"Eva"**.
✅ Add **5 bonus marks** to all students and print the updated Series.
✅ Display only the students who scored **above 90** after the bonus.
✅ Calculate and print the **average marks** of the class (after the bonus).

```python
import pandas as pd

# Sample dictionary containing student names and their initial scores
student_scores = {
    "Alice": 85,
    "Bob": 92,
    "Charlie": 78,
    "David": 88,
    "Eva": 95
}

# 1. Create a Pandas Series named 'scores'
scores = pd.Series(student_scores)

# 2. Print the marks scored by "Charlie" and "Eva"
print("--- Marks for Charlie and Eva ---")
print(f"Charlie: {scores['Charlie']}")
print(f"Eva: {scores['Eva']}\n")

# 3. Add 5 bonus marks to all students and print the updated Series
scores = scores + 5
print("--- Updated Scores (with +5 Bonus Marks) ---")
print(scores)
print()

# 4. Display only the students who scored above 90 after the bonus
print("--- Students Scoring Above 90 After Bonus ---")
print(scores[scores > 90])
print()

# 5. Calculate and print the average marks of the class
average_marks = scores.mean()
print(f"--- Class Average (After Bonus) ---")
print(f"Average Marks: {average_marks}")
```

🧠 Python Pandas Logical Practice Task
You are given three Series representing student data:
Names of students
Their ages
Their marks
These series are combined into a DataFrame as shown earlier.
🎯 Your Task:
Using the existing DataFrame, perform the following operations:
✅ Print the details (name, age, marks) of the student who scored the highest marks.
✅ Add a new student "Tharushi" with age 23 and marks 88 using the index "Th".
✅ Display the details of students who are 25 years old or older.
✅ Increase the marks of students below 80 by 10 bonus points.
✅ Calculate and print the average marks of all students after the bonus.


```python
import pandas as pd
name = pd.Series(["Alice","Bob","Charlie","David"])
age = pd.Series([22,21,20,21])
marks = pd.Series([95,98,93,92])
index_labels = pd.Series(["Al","Bb","ch","Dd"])
df = pd.DataFrame({"Name": name, "Age": age, "Marks": marks})
df.index = index_labels
df
highest_student = df.loc[df['Marks'].idxmax()]
print("--- Highest Scoring Student ---")
print(highest_student)
print()
df.loc["Th"] = ["Tharushi", 23, 88]
print("--- DataFrame After Adding Tharushi ---")
print(df)
print()
older_students = df[df['Age'] >= 25]
print("--- Students 25 or Older ---")
print(older_students)
print()
df.loc[df['Marks'] < 80, 'Marks'] += 10
print("--- DataFrame After Applying Bonus ---")
print(df)
print()

# ✅ 5. Calculate and print the average marks after the bonus
average_marks = df['Marks'].mean()
print(f"Average Marks of all students: {average_marks:.2f}")
```
